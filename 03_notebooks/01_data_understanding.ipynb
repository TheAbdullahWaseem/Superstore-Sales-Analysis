###  2.4 — Missing Values

#####Identifying columns with missing data and what percentage of that column is missing. Missing values must be handled before any analysis otherwise calculations will be incorrect.
---

```python
print("missing Values Analysis")
print("-" * 40)

missing_count = df.isna().sum()
missing_pct = (missing_count/ len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing %": missing_pct.round(2)
})

missing_with_data = missing_df[missing_df["Missing Count"] > 0]

if len(missing_with_data) == 0:
  print("No missing values found in any column")
else:
  print(missing_with_data)

print(f"\nTotal missing values: {df.isna().sum().sum()}")
```

### 2.5 — Duplicate Rows

#####Checking for completely duplicate rows in the dataset. Duplicates inflate counts and totals and must be identified before cleaning.
---

```python
print("Duplicate Rows Check:")
print("-" * 40)

duplicates = df.duplicated().sum()

if duplicates == 0:
    print("No duplicate rows found.")
else:
    print(f"Duplicate rows found: {duplicates}")
    print(f"Percentage of data: {(duplicates / len(df)) * 100:.2f}%")
```

### 2.7 — Categorical Column Breakdown

###  2.8 — Date Range

#####Understanding the time period covered by the dataset. We check the earliest and latest order dates and confirm which years are included.
---

```python
print("Date Range Analysis:")
print("-" * 40)

df['Order date'] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

earliest = df["Order Date"].min()
latest = df["Order Date"].max()

years = sorted(df['Order Date'].dt.year.unique())
date_range_days = (latest - earliest).days

print(f"Earliest Order  : {earliest.strftime('%B %d, %Y')}")
print(f"Latest Order    : {latest.strftime('%B %d, %Y')}")
print(f"Total Date Span : {date_range_days} days ({date_range_days//365} years)")
print(f"Years Covered   : {years}")
```

### 2.9 — Financial Overview

#####A high level summary of the key financial figures in the dataset. This gives us our first glimpse of the business performance before any deep analysis begins.
---

```python
print("Financial Overview:")
print("=" * 40)

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
profit_margin = (total_profit / total_sales) * 100
avg_discount = df["Discount"].mean() * 100
max_discount = df["Discount"].max() * 100
min_profit = df["Profit"].min()
max_profit = df["Profit"].max()
avg_shipping = df["Shipping Cost"].mean()
unique_customers = df["Customer Name"].nunique()
total_orders = df["Order ID"].nunique()

print(f"Total Sales          : ${total_sales:>12,.2f}")
print(f"Total Profit         : ${total_profit:>12,.2f}")
print(f"Profit Margin        : {profit_margin:>11.2f}%")
print(f"Avg Discount         : {avg_discount:>11.1f}%")
print(f"Max Discount         : {max_discount:>11.1f}%")
print(f"Min Profit (loss)    : ${min_profit:>12,.2f}")
print(f"Max Profit           : ${max_profit:>12,.2f}")
print(f"Avg Shipping Cost    : ${avg_shipping:>12,.2f}")
print(f"Unique Customers     : {unique_customers:>12,}")
print(f"Total Unique Orders  : {total_orders:>12,}")
```
